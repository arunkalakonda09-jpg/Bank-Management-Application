```python
%%writefile main.cpp
#include <iostream>
#include <fstream>
#include <cstring>
#include <limits>

using namespace std;

const char FILE_NAME[] = "students.dat";

class Student {
private:
    int rollNo;
    char name[50];
    char course[50];
    float marks;

public:
    Student() {
        rollNo = 0;
        strcpy(name, "");
        strcpy(course, "");
        marks = 0.0;
    }

    int getRollNo() const {
        return rollNo;
    }

    void input() {
        cout << "Enter Roll Number: ";
        cin >> rollNo;
        cin.ignore(numeric_limits<streamsize>::max(), '\n');

        cout << "Enter Name: ";
        cin.getline(name, 50);

        cout << "Enter Course: ";
        cin.getline(course, 50);

        cout << "Enter Marks: ";
        cin >> marks;
    }

    void display() const {
        cout << "\nRoll Number : " << rollNo;
        cout << "\nName        : " << name;
        cout << "\nCourse      : " << course;
        cout << "\nMarks       : " << marks;
        cout << "\n----------------------------";
    }

    void updateDetails() {
        cin.ignore(numeric_limits<streamsize>::max(), '\n');

        cout << "Enter New Name: ";
        cin.getline(name, 50);

        cout << "Enter New Course: ";
        cin.getline(course, 50);

        cout << "Enter New Marks: ";
        cin >> marks;
    }
};

void addStudent() {
    Student student;
    student.input();

    ofstream file(FILE_NAME, ios::binary | ios::app);

    if (!file) {
        cout << "\nError: Unable to open file.\n";
        return;
    }

    file.write(reinterpret_cast<char*>(&student), sizeof(Student));
    file.close();

    cout << "\nStudent record added successfully.\n";
}

void displayStudents() {
    Student student;
    ifstream file(FILE_NAME, ios::binary);

    if (!file) {
        cout << "\nNo student records found.\n";
        return;
    }

    cout << "\n========= STUDENT RECORDS =========";

    while (file.read(reinterpret_cast<char*>(&student), sizeof(Student))) {
        student.display();
    }

    file.close();
}

int searchStudent(int rollNo, Student &student) {
    ifstream file(FILE_NAME, ios::binary);

    if (!file) {
        return -1;
    }

    int position = 0;

    while (file.read(reinterpret_cast<char*>(&student), sizeof(Student))) {
        if (student.getRollNo() == rollNo) {
            file.close();
            return position;
        }
        position++;
    }

    file.close();
    return -1;
}

void searchStudentRecord() {
    int rollNo;
    Student student;

    cout << "Enter Roll Number to search: ";
    cin >> rollNo;

    int position = searchStudent(rollNo, student);

    if (position == -1) {
        cout << "\nStudent record not found.\n";
    } else {
        cout << "\nStudent found:";
        student.display();
    }
}

void updateStudent() {
    int rollNo;
    Student student;

    cout << "Enter Roll Number to update: ";
    cin >> rollNo;

    fstream file(FILE_NAME, ios::binary | ios::in | ios::out);

    if (!file) {
        cout << "\nNo student records found.\n";
        return;
    }

    bool found = false;

    while (file.read(reinterpret_cast<char*>(&student), sizeof(Student))) {
        if (student.getRollNo() == rollNo) {
            cout << "\nCurrent Record:";
            student.display();

            cout << "\nEnter updated details:\n";
            student.updateDetails();

            file.seekp(-static_cast<streamoff>(sizeof(Student)), ios::cur);
            file.write(reinterpret_cast<char*>(&student), sizeof(Student));

            found = true;
            break;
        }
    }

    file.close();

    if (found) {
        cout << "\nStudent record updated successfully.\n";
    } else {
        cout << "\nStudent record not found.\n";
    }
}

void deleteStudent() {
    int rollNo;
    Student student;

    cout << "Enter Roll Number to delete: ";
    cin >> rollNo;

    ifstream inFile(FILE_NAME, ios::binary);

    if (!inFile) {
        cout << "\nNo student records found.\n";
        return;
    }

    ofstream tempFile("temp.dat", ios::binary);
    bool found = false;

    while (inFile.read(reinterpret_cast<char*>(&student), sizeof(Student))) {
        if (student.getRollNo() != rollNo) {
            tempFile.write(reinterpret_cast<char*>(&student), sizeof(Student));
        } else {
            found = true;
        }
    }

    inFile.close();
    tempFile.close();

    remove(FILE_NAME);
    rename("temp.dat", FILE_NAME);

    if (found) {
        cout << "\nStudent record deleted successfully.\n";
    } else {
        cout << "\nStudent record not found.\n";
    }
}

int main() {
    int choice;

    do {
        cout << "\n========== STUDENT MANAGEMENT SYSTEM ==========\n";
        cout << "1. Add Student\n";
        cout << "2. Display All Students\n";
        cout << "3. Search Student\n";
        cout << "4. Update Student\n";
        cout << "5. Delete Student\n";
        cout << "6. Exit\n";
        cout << "Enter your choice: ";
        cin >> choice;

        if (cin.fail()) {
            cin.clear();
            cin.ignore(numeric_limits<streamsize>::max(), '\n');
            cout << "\nInvalid input. Please enter a number.\n";
            continue;
        }

        switch (choice) {
            case 1:
                addStudent();
                break;

            case 2:
                displayStudents();
                break;

            case 3:
                searchStudentRecord();
                break;

            case 4:
                updateStudent();
                break;

            case 5:
                deleteStudent();
                break;

            case 6:
                cout << "\nExiting Student Management System...\n";
                break;

            default:
                cout << "\nInvalid choice. Please try again.\n";
        }

    } while (choice != 6);

    return 0;
}
```
